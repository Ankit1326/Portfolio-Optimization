```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.optimize import minimize
```

```python
num_assets = len(returns.columns)
mean_returns = returns.mean() * 252
cov_matrix = returns.cov() * 252
risk_free = risk_free_rate
print("Number of Assets :", num_assets)
```

```python
equal_weights = np.repeat(1/num_assets, num_assets)

equal_return = np.sum(mean_returns * equal_weights)

equal_volatility = np.sqrt(
    np.dot(equal_weights.T,
           np.dot(cov_matrix, equal_weights))
)
equal_sharpe = (equal_return-risk_free)/equal_volatility
print("Equal Weight Portfolio")
print("Return :", round(equal_return,4))
print("Volatility :", round(equal_volatility,4))
print("Sharpe :", round(equal_sharpe,4))
```

```python
allocation = pd.DataFrame({

    "Stock":returns.columns,

    "Weight":equal_weights

})
allocation.head(20)
```

```python
plt.figure(figsize=(4,4))
plt.pie(
    equal_weights,
    labels=None,
    startangle=90
)
plt.title("Equal Weight Portfolio")
plt.show()
```

```python
num_portfolios = 50000
results = np.zeros((3,num_portfolios))
weights_record = []
```

```python
for i in range(num_portfolios):

    weights = np.random.random(num_assets)

    weights /= np.sum(weights)

    weights_record.append(weights)

    portfolio_return = np.sum(mean_returns * weights)

    portfolio_volatility = np.sqrt(

        np.dot(weights.T,

        np.dot(cov_matrix,weights))

    )
    sharpe = (portfolio_return-risk_free)/portfolio_volatility
    results[0,i] = portfolio_return
    results[1,i] = portfolio_volatility
    results[2,i] = sharpe
```

```python
mc_results = pd.DataFrame({

    "Return":results[0],

    "Risk":results[1],

    "Sharpe":results[2]

})

mc_results.head()
```

```python
best = mc_results.iloc[mc_results["Sharpe"].idxmax()]

print(best)
```

```python
minimum = mc_results.iloc[mc_results["Risk"].idxmin()]
print(minimum)
```

```python
plt.figure(figsize=(6,4))
scatter = plt.scatter(
    mc_results["Risk"],
    mc_results["Return"],
    c=mc_results["Sharpe"],
    cmap="viridis",
    alpha=.4)
plt.colorbar(scatter,label="Sharpe Ratio")
plt.scatter(
    best["Risk"],
    best["Return"],
    color="red",
    marker="*",
    s=100,
    label="Maximum Sharpe")
plt.scatter(
    minimum["Risk"],
    minimum["Return"],
    color="blue",
    marker="D",
    s=100,
    label="Minimum Risk")
plt.xlabel("Annual Volatility")
plt.ylabel("Annual Return")
plt.title("Monte Carlo Efficient Frontier")
plt.legend()
plt.show()
```
